```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/BMW sales data (2010-2024).csv")
ventas_por_modelo = df.groupby("Model")["Sales_Volume"].sum()

plt.bar(ventas_por_modelo.index, ventas_por_modelo.values)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df_wine = pd.read_csv('/content/wine_quality_merged.csv')
calidad_por_tipo = df_wine.groupby('type')['quality'].mean()

plt.bar(calidad_por_tipo.index, calidad_por_tipo.values)
plt.title('Promedio de Calidad por Tipo de Vino')
plt.xlabel('Tipo de Vino')
plt.ylabel('Calidad Promedio')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df_biomes = pd.read_csv('/content/Biomes.csv')
conteo_biomas = df_biomes['name'].value_counts()

plt.bar(conteo_biomas.index, conteo_biomas.values)
plt.title('Distribución de Biomas')
plt.xlabel('Bioma')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.show()
```
